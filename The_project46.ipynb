```python
!pip install rdkit
from rdkit import Chem
from rdkit import RDLogger
RDLogger.DisableLog("rdApp.*")
leaving_groups = {
    "Br",
    "Cl",
    "I"}
cpds=input("Enter the molecule: ")
mol=Chem.MolFromSmiles(cpds)
if mol is None:
    print("Invalid SMILES.")
    exit()
def find_substrate_carbon(mo):
    try:
       atoms=mo.GetAtoms()
       for atom in atoms:
        if atom.GetSymbol() in leaving_groups:
            for neighbors in  atom.GetNeighbors():
                if neighbors.GetSymbol()=="C":
                    return neighbors
    except AttributeError:
        return None
if mol is not None:
    atomos=find_substrate_carbon(mol)
def classify_carbon(atom):
    carbon_count = 0
    for a in atom.GetNeighbors():
        if a.GetSymbol()=="C":
            carbon_count=carbon_count + 1
    if carbon_count ==1:
        return "primary"
    elif carbon_count ==2:
        return "Secondary"
    elif carbon_count ==3:
        return "Tertariy"
if mol is not None:
  substrate_type=classify_carbon(atomos)
nucleophiles_data = {
    "CN": {
        "strength": "strong",
        "charge": -1,
        "bulky": False},
"OH": {
        "strength": "strong",
        "charge": -1,
        "bulky": False},
"O": {
        "strength": "weak",
        "charge": 0,
        "bulky": False}}
nucleophiles=input("Enter the nucleophile : ")
def classify_nucleophile(nucleophile):
    if nucleophile in nucleophiles_data .keys():
        return nucleophiles_data[nucleophile]
    return None
classfied_nuceli=classify_nucleophile(nucleophiles)
if classfied_nuceli is None:
    print("Unknown nucleophile.")
    exit()
def is_sn2_favorable(nuceli,carbon_type):
    if nuceli["strength"]=="strong"  and carbon_type=="primary":
        return "SN2"
    elif nuceli["strength"]=="strong" and carbon_type=="Secondary":
        return "SN2"
    return None
if mol is not None:
   reaction=is_sn2_favorable(classfied_nuceli,substrate_type)
import csv
with open("/content/sample_data/solution.csv","w") as file:
    s=file.write("name,polarity,type,toxicity\n")
    e=file.write("DMSO,high,aprotic,low\n")
    d=file.write("DMF,high,aprotic,high\n")
    t=file.write("Hexane,low,aprotic,medium\n")
with open("/content/sample_data/solution.csv","a") as file:
    m=file.write("H2O,high,protic,low")
def load_solvents() :
    with open("/content/sample_data/solution.csv","r") as file:
        sol_data=[]
        reader=csv.DictReader(file)
        for r in reader:
          sol_data.append(r)
        return sol_data
solvents=load_solvents()
def load_reaction_rule():
    reaction_rule={"SN2":{
        "polarity":"high",
        "type":"aprotic"},
    "E1":{"polarity":"high",
        "type":"protic"}}
    return reaction_rule
reaction_rules=load_reaction_rule()
if mol is not None:
 print("toxcity level label")
 print("low-1,medium-2,high-3")

def rule_solvent_connector(sol,rule,reaction): #takes all the rxn from reaction_rules and compares it with load_solvents.
    sol_rec=[]
    for solvent in sol:
            if solvent["polarity"]==rule[reaction]["polarity"] and solvent["type"]==rule[reaction]["type"]:
                sol_rec.append(solvent)
    return sol_rec
if mol is not None:
  if reaction in reaction_rules:
    recs=rule_solvent_connector(solvents,reaction_rules,reaction)
  else:
    print("We do not have the reaction!")
for solve in solvents:
      if solve["toxicity"]=="high": # high toxicity.
        solve["toxicity"]=3
      elif solve["toxicity"]=="medium": # mild toxicity.
        solve["toxicity"]=2
      elif solve["toxicity"]=="low": # not toxic to some extent.
          solve["toxicity"]=1
def sort_recommendations(rec):
     us=sorted(rec,key=lambda x: x["toxicity"])
     return us
if mol is not None:
   y=sort_recommendations(recs)
def display_results(recommendations):
        print("Recommended solvents:")
        for w in recommendations:
            for k,v in w.items():
                   print(f'{k}:{v}')
            print("="*30)
if mol is not None:
 if reaction in reaction_rules.keys():
    display_results(y)
```
